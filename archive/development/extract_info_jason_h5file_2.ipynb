```python
# function to find datasets with peaks
def find_datasets_with_peaks(file_path: Path) -> list[str]:
    """Find datasets with peaks in the HDF5 file.

    Args:
        file_path (Path): Path to the HDF5 file.

    Returns:
        list[str]: List of dataset IDs with peaks.
    """
    expts_with_peaks = {}
    with h5py.File(file_path, 'r') as fp:
        for expt_id in fp['JasonDocument/NMR/NMRData']:
            try:
                peakList = fp[f'JasonDocument/NMR/NMRData/{expt_id}/Peaks/PeakList']
                print(len(peakList), "peaks found in", expt_id)
                if len(peakList) > 0:
                    expts_with_peaks[expt_id] = {}
                    expts_with_peaks[expt_id]['spec_info'] = {}
                    expts_with_peaks[expt_id]['peaks'] = {}
                    expts_with_peaks[expt_id]['integrals'] = {}
                    expts_with_peaks[expt_id]['multiplets'] = {}
            except KeyError:
                print(f"No peaks found in {expt_id}")
                continue
    return expts_with_peaks 

            
        
```

```python
def find_expt_details( fn, datasets):
    """Find experiment details for the given datasets.

    Args:
        fn (Path): Path to the HDF5 file.
        datasets (dict): Dictionary of datasets with peaks.

    Returns:
        dict: Dictionary of experiment details.
    """
    
    with h5py.File(fn, 'r') as fp:
        for expt_id in datasets.keys():
            specInfo = datasets[expt_id]["spec_info"]
            try:
                spec_info = fp[f'JasonDocument/NMR/NMRData/{expt_id}/SpecInfo']
                nucleides = fp[f'JasonDocument/NMR/NMRData/{expt_id}/SpecInfo/Nucleides']

                specInfo['pulsesequence'] = spec_info.attrs.get('PulseProgram', '').decode()
                specInfo['experimenttype'] = spec_info.attrs.get('ExperimentType.str', '').decode()
                specInfo['datafilename'] = spec_info.attrs.get('OrigFilename.filepath.str', '').decode()
                specInfo['solvent'] = spec_info.attrs.get('Solvent', '')
                specInfo['specfrequency'] = [sf for sf in spec_info.attrs.get('SpectrometerFrequencies', []) if sf > 0]
                specInfo['type'] = str(len(specInfo['specfrequency']))+'D'  # Number of frequencies
                specInfo['temperature'] = spec_info.attrs.get('Temperature', None)

                Isotopes = [nucleides[n].attrs["Isotope"] for n in list(nucleides)]
                Names = [nucleides[n].attrs["Name"].decode() for n in list(nucleides)]
                nuclei = [ f"{i}{n}" for n, i in zip(Names, Isotopes) if i > 0]
                if len(nuclei) == 1:
                    specInfo['nucleus'] = nuclei[0]
                else:
                    specInfo['nucleus'] = nuclei

                # datasets[expt_id]["specInfo"] = specInfo
            except KeyError:
                print(f"KeyError: SpecInfo not found for experiment {expt_id}")
                continue
    return datasets
```

### Extract peaks from jason h5 file

```python

   
def get_peaklist(fp, expt_id):
    try:
        # f['JasonDocument/NMR/NMRData/1/Peaks/PeakList']
        return list(fp[f'JasonDocument/NMR/NMRData/{expt_id}/Peaks/PeakList'])
    except KeyError:
        return []
    
def get_peakinfo(fp, expt_id, pk_id):
    peak_vals = {}
    try:
        # f['JasonDocument/NMR/NMRData/1/Peaks/PeakList/0']
        peak_group = fp[f'JasonDocument/NMR/NMRData/{expt_id}/Peaks/PeakList/{pk_id}']
        peak_vals['intensity'] = peak_group.attrs.get('Height', None)
        peak_vals['F2_ppm'] = peak_group.attrs.get('Pos', None)[0]
        peak_vals['F1_ppm'] = peak_group.attrs.get('Pos', None)[1]
        return peak_vals
    except KeyError:
        return {}
```

```python
with h5py.File(fn, 'r') as f:
    # extract peaks from spectra
    for expt in datasets_with_peaks.keys():
        for pk_id in get_peaklist(f, expt):
            peak_info = get_peakinfo(f, expt, pk_id)
            datasets_with_peaks[expt]['peaks'][pk_id] = peak_info
```

```python
def get_integrallist(fp, expt_id):
    try:
        return list(fp[f'JasonDocument/NMR/NMRData/{expt_id}/Multiplets_Integrals/MultipletList'])
    except KeyError:
        return []

def get_integralinfo(fp, expt_id, int_id):
    integral_vals = {}
    try:
        # f['JasonDocument/NMR/NMRData/1/Multiplets_Integrals/MultipletList/0']
        integral_group = fp[f'JasonDocument/NMR/NMRData/{expt_id}/Multiplets_Integrals/MultipletList/{int_id}']
        integral_vals['intensity'] = integral_group.attrs.get('Value', None)
        integral_vals['F2_ppm_range'] = integral_group.attrs.get('SpectrumRange[0]', None)
        integral_vals['F1_ppm_range'] = integral_group.attrs.get('SpectrumRange[1]', None)
        return integral_vals
    except KeyError:
        return {}
```

```python
# get integrals
with h5py.File(fn, 'r') as f:
    # extract integrals from spectra
    for expt in datasets_with_peaks.keys():
        for int_id in get_integrallist(f, expt):
            integral_info = get_integralinfo(f, expt, int_id)
            datasets_with_peaks[expt]['integrals'][int_id] = integral_info
```
